```python
import numpy as np
import matplotlib.pyplot as plt

# Data
base_path = '/Users/songsifan/Library/CloudStorage/OneDrive-MassGeneralBrigham/Scripts/OD_202401_DOC/OD_Writing/202503_ICCV25_OWT/bak_figs/'

organs = ['Liver', 'Kidney', 'Spleen', 'Pancreas', 'Avg']
model1_scores = [96.81,94.84,95.59,78.23,91.37]
model2_scores = [97.09,94.83,95.92,78.08,91.43]
save_path = base_path + 'abd1k_3d.png'

# organs = ['Liver', 'Kidney', 'Spleen', 'Pancreas', 'Avg']
# model1_scores = [96.65,93.15,93.96,81.87,91.41]
# model2_scores = [96.85,93.53,94.22,82.22,91.70]
# save_path = base_path + 'autopet_3d.png'

# organs = ['Liver', 'Kidney', 'Spleen', 'Pancreas', 'Avg']
# model1_scores = [94.72,92.61,92.23,71.33,87.72]
# model2_scores = [94.76,92.35,92.24,71.54,87.65]
# save_path = base_path + 'delay_3d.png'

# organs = ['Liver', 'Kidney', 'Spleen', 'Pancreas', 'Avg']
# model1_scores = [94.74,92.18,92.14,70.17,87.31]
# model2_scores = [94.92,92.06,92.04,70.54,87.34]
# save_path = base_path + 'PreArtery_3d.png'

# Set up bar positions
x = np.arange(len(organs))
width = 0.35  # Width of bars

# Create figure and axis
fig, ax = plt.subplots(figsize=(6, 4))

# Create bars
rects1 = ax.bar(x - width/2, model1_scores, width, label='OWT Direct Seg.', color='#C5E0B4', alpha=1.0)
rects2 = ax.bar(x + width/2, model2_scores, width, label='OWT Indirect Seg.', color='#FAAC9A', alpha=1.0)

# rects1 = ax.bar(x - width/2, model1_scores, width, label='OWT Direct Seg.', color='#2F6669', alpha=1.0)
# rects2 = ax.bar(x + width/2, model2_scores, width, label='OWT Indirect Seg.', color='#FE925F', alpha=1.0)

# Customize plot
# ax.set_ylabel('Dice Score (%)', fontsize=12, fontweight='bold')
# ax.set_xlabel('Organs', fontsize=12, fontweight='bold')
# ax.set_title('Organ Segmentation Performance Comparison', fontsize=14, fontweight='bold', pad=20)
ax.set_xticks(x)
ax.set_xticklabels(organs, fontsize=10, rotation=0)
ax.legend(fontsize=10)

# Set y-axis range
ax.set_ylim(60, 100)

# Add value labels on top of bars
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(rect.get_x() + rect.get_width()/2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom',
                    fontsize=9)

autolabel(rects1)
autolabel(rects2)

# Add grid lines
ax.grid(True, linestyle='--', alpha=0.7, axis='y')

# Adjust layout
plt.tight_layout()

# Save plot
plt.savefig(save_path, dpi=300, bbox_inches='tight')

# Show plot
plt.show()


```
