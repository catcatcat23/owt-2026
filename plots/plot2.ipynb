```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create the data
data = {
    'Tokens': ['5', '10', '20', '40'],
    'Whole Image L2': [2.72E-04, 1.79E-04, 1.49E-04, 1.07E-04],
    'Whole Image LPIPS': [0.0367, 0.0264, 0.0232, 0.0169],
    'Whole Image SSIM-3D': [0.9775, 0.9861, 0.9881, 0.9913],
    '1 Token Group L2': [3.63E-04, 3.31E-04, 3.32E-04, 2.94E-04],
    '1 Token Group LPIPS': [0.0077, 0.0071, 0.0071, 0.0064],
    '1 Token Group SSIM-3D': [0.9747, 0.9773, 0.9778, 0.9824],
    'Seg Direct': [0.8789, 0.8938, 0.8952, 0.9075],
    'Seg Indirect': [0.8722, 0.8967, 0.9014, 0.9073]
}

colors = ['#3B93C3', '#DE8F06']

df = pd.DataFrame(data)

# Create figure and axis objects with subplots()
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(11, 7))

# Plot L2 metrics
line1 = ax1.plot(df['Tokens'], df['Whole Image L2'], marker='o', label='Whole Image Rec.', color=colors[0])
line2 = ax1.plot(df['Tokens'], df['1 Token Group L2'], marker='s', label='1 Token Group Rec.', color=colors[1])
for i, val in enumerate(df['Whole Image L2']):
    ax1.annotate(f'{val:.2e}', (df['Tokens'][i], val), textcoords="offset points", xytext=(0,10), ha='center')
for i, val in enumerate(df['1 Token Group L2']):
    ax1.annotate(f'{val:.2e}', (df['Tokens'][i], val), textcoords="offset points", xytext=(0,-10), ha='center')
ax1.set_ylabel('L2', fontsize=13)
ax1.legend(fontsize=13)
ax1.grid(True, color='lightgray', linestyle='--')

# Plot LPIPS metrics
ax2.plot(df['Tokens'], df['Whole Image LPIPS'], marker='o', label='Whole Image Rec.', color=colors[0])
ax2.plot(df['Tokens'], df['1 Token Group LPIPS'], marker='s', label='1 Token Group Rec.', color=colors[1])
for i, val in enumerate(df['Whole Image LPIPS']):
    ax2.annotate(f'{val:.4f}', (df['Tokens'][i], val), textcoords="offset points", xytext=(0,10), ha='center')
for i, val in enumerate(df['1 Token Group LPIPS']):
    ax2.annotate(f'{val:.4f}', (df['Tokens'][i], val), textcoords="offset points", xytext=(0,-10), ha='center')
ax2.set_ylabel('LPIPS', fontsize=13)
ax2.legend(fontsize=13)
ax2.grid(True, color='lightgray', linestyle='--')

# Plot SSIM-3D metrics
ax3.plot(df['Tokens'], df['Whole Image SSIM-3D'], marker='o', label='Whole Image Rec.', color=colors[0])
ax3.plot(df['Tokens'], df['1 Token Group SSIM-3D'], marker='s', label='1 Token Group Rec.', color=colors[1])
for i, val in enumerate(df['Whole Image SSIM-3D']):
    ax3.annotate(f'{val:.4f}', (df['Tokens'][i], val), textcoords="offset points", xytext=(0,10), ha='center')
for i, val in enumerate(df['1 Token Group SSIM-3D']):
    ax3.annotate(f'{val:.4f}', (df['Tokens'][i], val), textcoords="offset points", xytext=(0,-10), ha='center')
ax3.set_xlabel('Number of Tokens', fontsize=13)
ax3.set_ylabel('SSIM-3D', fontsize=13)
ax3.legend(fontsize=13)
ax3.grid(True, color='lightgray', linestyle='--')

# Plot Seg Direct metrics
ax4.plot(df['Tokens'], df['Seg Direct'], marker='o', label='1 Token Group Seg.', color=colors[1])
for i, val in enumerate(df['Seg Direct']):
    ax4.annotate(f'{val:.4f}', (df['Tokens'][i], val), textcoords="offset points", xytext=(0,10), ha='center')
ax4.set_xlabel('Number of Tokens', fontsize=13)
ax4.set_ylabel('Dice', fontsize=13)
ax4.legend(fontsize=13)
ax4.grid(True, color='lightgray', linestyle='--')

plt.tight_layout()

# Save the figure
base_path = '/Users/songsifan/Library/CloudStorage/OneDrive-MassGeneralBrigham/Scripts/OD_202401_DOC/OD_Writing/202503_ICCV25_OWT/'
plt.savefig(base_path + 'Figure_Ablation_Token_Number.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Create figure and axes
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 3))

# Prepare data
autopet_df = pd.DataFrame({
    'Model': ['MAE', 'MAE/LPIPS', 'OWT', 'OWT/CNN'],
    'L2': [1.49E-04, 1.49E-04, 1.49E-04, 6.89E-05],
    'LPIPS': [0.0232, 0.0232, 0.0232, 0.0073]
})

face_df = pd.DataFrame({
    'Model': ['MAE', 'MAE/LPIPS', 'OWT', 'OWT/CNN'],
    'L2': [3.20E-03, 3.20E-03, 3.20E-03, 1.80E-03],
    'LPIPS': [0.1101, 0.1101, 0.1101, 0.0540]
})

# Set up colors and dimensions
colors = ['#2ecc71', '#3498db']
y = np.arange(len(autopet_df['Model']))  # Remove the [::-1] to keep original order
height = 0.35

# Plot L2 metrics
ax1.barh(y - height/2, autopet_df['L2'], height, label='AutoPET', color=colors[0])
ax1.barh(y + height/2, face_df['L2'], height, label='Face', color=colors[1])
ax1.set_xlabel('L2')
ax1.set_title('L2 Comparison')
ax1.set_yticks(y)
ax1.set_yticklabels(autopet_df['Model'])
ax1.legend()
ax1.grid(True)

# Plot LPIPS metrics
ax2.barh(y - height/2, autopet_df['LPIPS'], height, label='AutoPET', color=colors[0])
ax2.barh(y + height/2, face_df['LPIPS'], height, label='Face', color=colors[1])
ax2.set_xlabel('LPIPS')
ax2.set_title('LPIPS Comparison')
ax2.set_yticks(y)
ax2.set_yticklabels(autopet_df['Model'])
ax2.legend()
ax2.grid(True)

plt.tight_layout()

# Save the figure
plt.savefig(base_path + 'Figure_AutoPET_Face_Metrics_Comparison.png', dpi=300, bbox_inches='tight')
plt.show()
```
